```python
#A
import pandas as pd
dates=pd.to_datetime([
    '2026-01-01',
    '2026-01-02',
    '2026-01-03',
    '2026-01-04',
    '2026-01-05'
])
temperature=[28,30,29,31,32]
time_series=pd.Series(temperature,index=dates)
print("Time Series:")
print(time_series)
```

```python
#B
import pandas as pd
date_index=pd.date_range(start='2026-01-01',periods=7,freq='D')
print("Generated DatatimeIndex:")
print(date_index)
```

```python
#C
import pandas as pd
india_dates=pd.date_range(start='2026-01-01 09:00',periods=3,freq='h',tz='Asia/Kolkata')
print("1. Date Range with Asia/Kolkata Time Zone:")
print(india_dates)
dates=pd.date_range(start='2026-01-01 09:00',periods=3,freq='h')
print("\n2. TimeZone-naive Data Range:")
print(dates)
localized_dates=dates.tz_localize('Asia/Kolkata')
print("\n3. After Localizing to Asia/Kolkata:")
print(localized_dates)
new_york_dates=localized_dates.tz_convert('America/New_York')
print("\n4. Converted to America/New York:")
print(new_york_dates)
india_series=pd.Series([100,200,300],index=localized_dates)
new_york_series=pd.Series([400,500,600],index=new_york_dates)
print("\n5. India Time Series:")
print(india_series)
print("\n6. New York Time Series:")
print(new_york_series)
combined_series=pd.concat([india_series,new_york_series])
print("\n7. Combined Time Series:")
print(combined_series)
```

```python
#D
import pandas as pd
p=pd.Period('2026-01',freq='M')
print("Original Period:")
print(p)
print("\nAfter adding 2:")
print(p+2)
print("\nAfter adding 5:")
print(p+5)
print("\nAfter subtracting 1:")
print(p-1)
print("\nAfter subtracting 3:")
print(p-3)
periods=pd.period_range(start='2026-01',periods=6,freq='M')
print("\nPeriod Range:")
print(periods)
```

```python
#E
import pandas as pd
p=pd.Period('2026-01',freq='M')
print("Original Period:")
print(p)
daily_start=p.asfreq('D',how='start')
print("\nMonthly Period converted to Daily(start):")
print(daily_start)
daily_end=p.asfreq('D',how='end')
print("\nMonthly Period converted to Daily(end):")
print(daily_end)
period_index=pd.period_range(start='2026-01',periods=3,freq='M')
print("\nOriginal Period Index:")
print(period_index)
daily_period_index=period_index.asfreq('D',how='start')
print("\nPeriod Index converted to Daily Frequency:")
print(daily_period_index)
daily_period_index=period_index.asfreq('D',how='end')
print("\nPeriod Index converted to Daily Frequency(End):")
print(daily_period_index)
```

```python
#F
import pandas as pd
dates=pd.to_datetime([
    '2026-01-10',
    '2026-01-15',
    '2026-01-20',
    '2026-01-25',
])
sales=pd.Series([1000,1500,1800,2200],index=dates)
print("Original Sales Data:")
print(sales)
period_series=sales.to_period('M')
print("\nSales converted to Monthly Periods:")
print(period_series)
df=pd.DataFrame({
    'Sales':[1000,1500,1800,2200],
    'Profit':[200,300,400,500]
},index=dates)
print("\nOriginal DataFrame:")
print(df)
period_df=df.to_period('M')
print("\nDataFrame converted to Monthly Periods:")
print(period_df)
print_df=df.to_period('M')
print("\nDataFrame converted to Monthly Periods:")
print(print_df)
```

```python
#G
import pandas as pd
p = pd.Period('2026-01', freq='M')
print("Original Period:")
print(p)
daily_start = p.asfreq('D', how='start')
print("\nMonthly Period converted to Daily (Start):")
print(daily_start)
daily_end = p.asfreq('D', how='end')

print("\nMonthly Period converted to Daily (End):")
print(daily_end)
period_index = pd.period_range(
    start='2026-01',
    periods=3,
    freq='M'
)
print("\nOriginal PeriodIndex:")
print(period_index)
daily_period_index = period_index.asfreq(
    'D',
    how='start'
)
print("\nPeriodIndex converted to Daily Frequency:")
print(daily_period_index)
daily_period_index_end = period_index.asfreq(
    'D',
    how='end'
)
print("\nPeriodIndex converted to Daily Frequency (End):")
print(daily_period_index_end)
```
