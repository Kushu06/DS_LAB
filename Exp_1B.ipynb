```python
import pandas as pd
student_data = {
    "Roll_No":[11,12,13,14],
    "Name":["Kushvanth","Kartheek","Uday kiran","Ganesh"],
    "Department":["DS","DS","DS","DS"],
    "Percentage":[99,98,90,92]
}
df = pd.DataFrame(student_data)
df.to_json(
    "students.json",
    orient = "records",
    index = 4
)
print("JSON file created sucessfully")
```

```python
import pandas as pd
df = pd.read_json("students.json")
print("Students Data..")
print(df)
```
