```python
#5A
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[10,20,25,30]
plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.savefig("line_plot.png")
plt.show()
```

```python
#5B
import pandas as pd
df=pd.DataFrame({'category':['A','B','C'],'value':[10,20,15]})
df.plot(kind='bar',x='category',y='value')
plt.title("Bar Plot")
plt.show()
```

```python
#5C
import seaborn as sns
import numpy as np
data=np.random.rand(1000)
sns.histplot(data,kde=True)
plt.title("History Plot")
plt.show()
```

```python
#5D
x=np.random.rand(50)
y=np.random.rand(50)
plt.scatter(x,y)
plt.title("Scatter Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
#5E
sns.boxplot(data=np.random.rand(100,4))
plt.title("Box Plot")
plt.show()
```
