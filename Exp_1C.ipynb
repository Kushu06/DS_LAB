```python
import pandas as pd
student_data={
    "Roll_No":[11,12,13,14],
    "Name":["Kushvanth","Kartheek","Uday kiran","Ganesh"],
    "Department":["DS","DS","DS","DS"],
    "Percentage":[99,98,90,92]
}
course_data={
    "course_id":["c101","c102","c103","c104"],
    "course_name":["python","datascience","machinelearning","artificia inteligence"],
    "credits":[4,3,4,3]
}
student_df=pd.DataFrame(student_data)
course_df=pd.DataFrame(course_data)
with pd.ExcelWriter("college_data.xlsx",engine="openpyxl") as writer:
    student_df.to_excel(writer,sheet_name="Students",index=False)
    course_df.to_excel(writer,sheet_name="Courses",index=False)
print("multiple sheets written to college_data.xlsx")
```

```python
import pandas as pd
excel_data=pd.read_excel("college_data.xlsx",sheet_name=None)
print("data read from excel file:")
print("Available sheets:")
print(excel_data.keys())
print("\nstudent_sheet:")
print(excel_data["Students"])
print("\ncourse_sheet:")
print(excel_data["Courses"])
```
